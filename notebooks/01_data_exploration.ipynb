## Part 2 — Verified file roles & crop-level merge

Checking actual column content (crop lists, value ranges, percentage-sum patterns) against the originally assumed file roles turned up several corrections:

| File | Assumed role | Verified role |
|---|---|---|
| `sect1_hh_w1` | household + region identifier | Household **person roster** (age, sex, ...) + region codes. Region codes (`saq01`-`06`) actually exist identically in every PH file below, not just this one. |
| `sect1_ph_w1` | post-harvest crop list | **Not crop data** — no `crop_name` column at all. It's a person roster for the PH module respondent. |
| `sect_nr_ph_w1` | post-harvest non-roster details | Confirmed-ish: holder IDs for people *not* in the household roster. No crop/loss info. |
| `sect9_ph_w1` | crop disposition (sold/stored/lost) | **Storage** detail for field crops (cereals/pulses/oilseeds) — quantity/unit pairs, not disposition. |
| `sect10_ph_w1` | crop loss quantities + reasons | **Confirmed.** 9 lettered columns are 3 (occurred-flag, unit, quantity) triplets = 3 loss-reason slots per crop. |
| `sect11_ph_w1` | tree/root crops post-harvest | Field crops (same scope as sect9). Contains `q22_a`-`g`, percentages summing to ~100 — this is the **disposition breakdown** (sold/consumed/stored/lost shares). |
| `sect12_ph_w1` | storage type used | Garden/perennial crops (fruit, veg, spice). Same %-share disposition structure as sect11, for the garden-crop domain. This is the "tree/root/garden crop" file, not "storage type". |

**Grain mismatch:** `sect9`/`sect10` are one row per household-parcel-field-crop; `sect11`/`sect12` are one row per household-crop (already aggregated across parcels). `sect1_hh_w1` is one row per person. Merging all of these directly on `household_id` would fan out crop rows across household members. Plan:

1. Load only the 5 files that carry crop/loss/person signal: `sect9`, `sect10`, `sect11`, `sect12`, `sect1_hh`.
2. Rename columns to human-readable names — confirmed meanings get real names; everything still ambiguous keeps an honest generic name (`_unconfirmed`) rather than a guessed one, since inventing labels here would be actively misleading without the official LSMS codebook.
3. Aggregate `sect9` and `sect10` up to household+crop grain (sum across parcels/fields).
4. Concatenate `sect11` (field crops) + `sect12` (garden crops) into one disposition table.
5. Outer-merge storage + loss + disposition on `household_id` + `crop_code` + `crop_name`.
6. Attach household-level info (household size, rural flag, region codes) from `sect1_hh_w1`, aggregated to one row per household.

```python
# Geographic hierarchy (saq01-06) follows the standard published ERSS/LSMS-ISA
# Ethiopia admin coding (region > zone > woreda > town > kebele > ea). Kept as
# *_code, not decoded to names, since we don't have the region/zone code lookup.
GEO_RENAME = {
    "saq01": "region_code",
    "saq02": "zone_code",
    "saq03": "woreda_code",
    "saq04": "town_code",
    "saq05": "kebele_code",
    "saq06": "ea_code_local",
    "ea_id": "ea_id",
    "rural": "is_rural",
    "pw": "sampling_weight",
    "household_id": "household_id",
    "holder_id": "holder_id",
    "crop_name": "crop_name",
    "crop_code": "crop_code",
}

# --- sect9: storage detail (field crops) ---
# q01 is binary-like (stored y/n). q02-q05 and q08/q12/q13 are (quantity, unit)
# pairs but WHICH storage attribute each pair measures isn't decodable without
# the codebook, so they keep generic sequential names rather than guessed ones.
SECT9_RENAME = {
    **GEO_RENAME,
    "parcel_id": "parcel_id",
    "field_id": "field_id",
    "ph_s9q01": "storage_flag_unconfirmed",
    "ph_s9q02_a": "storage_qty1_unconfirmed", "ph_s9q02_b": "storage_qty1_unit_unconfirmed",
    "ph_s9q03_a": "storage_qty2_unconfirmed", "ph_s9q03_b": "storage_qty2_unit_unconfirmed",
    "ph_s9q04_a": "storage_qty3_unconfirmed", "ph_s9q04_b": "storage_qty3_unit_unconfirmed",
    "ph_s9q05_a": "storage_qty4_unconfirmed", "ph_s9q05_b": "storage_qty4_unit_unconfirmed",
    "ph_s9q06": "storage_flag2_unconfirmed",
    "ph_s9q07": "storage_flag3_unconfirmed",
    "ph_s9q08_a": "storage_qty5_unconfirmed", "ph_s9q08_b": "storage_qty5_unit_unconfirmed",
    "ph_s9q09": "storage_flag4_unconfirmed",
    "ph_s9q10": "storage_detail1_unconfirmed",
    "ph_s9q11": "storage_detail2_unconfirmed",
    "ph_s9q12_a": "storage_qty6_unconfirmed", "ph_s9q12_b": "storage_qty6_unit_unconfirmed",
    "ph_s9q13_a": "storage_qty7_unconfirmed", "ph_s9q13_b": "storage_qty7_unit_unconfirmed",
}

# --- sect10: loss reasons + quantities (all crop domains) --- CONFIRMED structure:
# 3 reason "slots", each a (occurred, unit, quantity) triplet. Which specific
# reason (pest/rain/theft/...) each slot represents is not decodable without
# the codebook, so slots are numbered, not named by cause.
SECT10_RENAME = {
    **GEO_RENAME,
    "parcel_id": "parcel_id",
    "field_id": "field_id",
    "crop_id": "crop_instance_id",
    "ph_s10q01_a": "loss_reason1_occurred", "ph_s10q01_b": "loss_reason1_unit", "ph_s10q01_c": "loss_reason1_qty",
    "ph_s10q01_d": "loss_reason2_occurred", "ph_s10q01_e": "loss_reason2_unit", "ph_s10q01_f": "loss_reason2_qty",
    "ph_s10q01_g": "loss_reason3_occurred", "ph_s10q01_h": "loss_reason3_unit", "ph_s10q01_i": "loss_reason3_qty",
}
# q02 (33 cols) and q03 (6 cols) blocks: pattern not confidently decoded, kept generic
for c in [f"ph_s10q02_{s}" for s in "abcdefghijklmnopqrstuvwxyz"] + [f"ph_s10q02_{s}a" for s in "klmnop"]:
    SECT10_RENAME[c] = f"loss_detail_{c.split('_')[-1]}_unconfirmed"
for s in "abcdef":
    SECT10_RENAME[f"ph_s10q03_{s}"] = f"loss_extra_{s}_unconfirmed"

# --- sect11 / sect12: disposition share breakdown ---
# q22 (sect11) / q19 (sect12), columns a-f are percentage shares of the crop's
# disposition that sum to ~100; the final letter (g) is a total/check column.
# Exact category per share (sold/consumed/seed/...) is not decodable without
# the codebook.
def disposition_rename(question_prefix):
    d = {**GEO_RENAME}
    letters = "abcdefg"
    for i, l in enumerate(letters[:6], start=1):
        d[f"{question_prefix}_{l}"] = f"disposition_share{i}_pct_unconfirmed"
    d[f"{question_prefix}_g"] = "disposition_total_check_pct"
    return d

SECT11_RENAME = {**GEO_RENAME, **disposition_rename("ph_s11q22")}
SECT12_RENAME = {**GEO_RENAME, **disposition_rename("ph_s12q19")}

# --- sect1_hh: household roster --- CONFIRMED: q00=person number, q03=sex, q04_a=age
SECT1_HH_RENAME = {
    **GEO_RENAME,
    "individual_id": "individual_id",
    "hh_s1q00": "person_number",
    "hh_s1q03": "sex",
    "hh_s1q04_a": "age",
}

df9 = pd.read_csv(DATA_DIR / "sect9_ph_w1.csv").rename(columns=SECT9_RENAME)
df10 = pd.read_csv(DATA_DIR / "sect10_ph_w1.csv").rename(columns=SECT10_RENAME)
df11 = pd.read_csv(DATA_DIR / "sect11_ph_w1.csv").rename(columns=SECT11_RENAME)
df12 = pd.read_csv(DATA_DIR / "sect12_ph_w1.csv").rename(columns=SECT12_RENAME)
df1_hh = pd.read_csv(DATA_DIR / "sect1_hh_w1.csv").rename(columns=SECT1_HH_RENAME)

df9.shape, df10.shape, df11.shape, df12.shape, df1_hh.shape
```

### Aggregate parcel/field-level files up to household + crop grain

`sect9` (storage) and `sect10` (loss) have one row per parcel/field within a crop; `sect11`/`sect12` (disposition) are already one row per household-crop. Sum the numeric quantity columns across parcels/fields so every table ends up at the same `household_id` + `crop_code` grain.

```python
GRAIN = ["household_id", "crop_code", "crop_name"]

# Identifier/geo/weight columns are numeric but must NOT be summed across
# parcels/fields (they're constant per household, not additive quantities).
NON_ADDITIVE_COLS = {
    "holder_id", "ea_id", "parcel_id", "field_id", "crop_instance_id",
    "region_code", "zone_code", "woreda_code", "town_code", "kebele_code",
    "ea_code_local", "is_rural", "sampling_weight",
}

def agg_to_crop_grain(df, grain=GRAIN):
    numeric_cols = df.select_dtypes("number").columns.difference(grain + list(NON_ADDITIVE_COLS))
    agg = df.groupby(grain, dropna=False)[list(numeric_cols)].sum(min_count=1)
    return agg.reset_index()

storage_agg = agg_to_crop_grain(df9)
loss_agg = agg_to_crop_grain(df10)

# Engineered convenience column: total quantity lost across the 3 reason slots.
# Safe to sum because all three are the same "qty" field by construction (verified
# structure), regardless of which specific reason each slot represents.
loss_agg["total_loss_qty"] = loss_agg[
    ["loss_reason1_qty", "loss_reason2_qty", "loss_reason3_qty"]
].sum(axis=1, min_count=1)

storage_agg.shape, loss_agg.shape
```

### Combine disposition tables (field crops + garden crops)

```python
df11["crop_domain"] = "field_crop"
df12["crop_domain"] = "garden_crop"

disposition_cols = GRAIN + ["crop_domain"] + [
    c for c in df11.columns if c.startswith("disposition_")
]
disposition_all = pd.concat(
    [df11[disposition_cols], df12[disposition_cols]],
    ignore_index=True,
)

# sect11/sect12 are already one row per household-crop, but guard against
# duplicate interviews by keeping the first record per household+crop.
disposition_all = disposition_all.drop_duplicates(subset=GRAIN, keep="first")

disposition_all.shape
```

### Household-level summary (from the person roster)

`sect1_hh_w1` is per-person; collapse it to one row per household (size + region/rural flags) before attaching it to the crop-level table.

```python
household_summary = (
    df1_hh.groupby("household_id")
    .agg(
        household_size=("person_number", "count"),
        region_code=("region_code", "first"),
        zone_code=("zone_code", "first"),
        woreda_code=("woreda_code", "first"),
        is_rural=("is_rural", "first"),
    )
    .reset_index()
)

household_summary.shape
```

### Final merge: one row per household + crop

```python
crop_loss_master = disposition_all.merge(
    storage_agg, on=GRAIN, how="outer", suffixes=("", "_storage")
).merge(
    loss_agg, on=GRAIN, how="outer", suffixes=("", "_loss")
).merge(
    household_summary, on="household_id", how="left"
)
crop_loss_master["survey_year"] = 2011

print(crop_loss_master.shape)
crop_loss_master.head()
```

```python
# Sanity checks
print("rows with a crop from disposition but no loss record:",
      crop_loss_master["total_loss_qty"].isna().sum())
print("rows with total_loss_qty > 0:", (crop_loss_master["total_loss_qty"] > 0).sum())
print("crop_domain counts:")
print(crop_loss_master["crop_domain"].value_counts(dropna=False))
crop_loss_master.describe(include="all").T
```

```python
OUT_PATH = Path("../data/processed/crop_loss_master_2011.csv")
crop_loss_master.to_csv(OUT_PATH, index=False)
print(f"saved {crop_loss_master.shape} to {OUT_PATH}")
```

## Caveats

- Columns suffixed `_unconfirmed` are structurally grouped (e.g. all 3 loss-reason triplets, all 6 disposition shares) but their exact real-world meaning (which reason, which disposition channel) is **not** verified — that requires the official LSMS-ISA Basic Information Document / questionnaire, which isn't in this repo yet.
- `region_code`/`zone_code`/`woreda_code` follow the standard published ERSS admin hierarchy convention but aren't decoded to actual place names (no lookup table available yet).
- `sect_nr_ph_w1` and `sect1_ph_w1` were intentionally excluded from the merge — they carry no crop or loss information (respondent-roster only).
- `crop_loss_master` is Wave 1 (2011-12) only. Repeat this pipeline per wave once Waves 2-5 are added to `data/raw/`, then concatenate before merging with CHIRPS rainfall.

## Part 3 — Wave 2 (2013-14)

Same pipeline, same 7 assumed file roles, applied to Wave 2. Two format differences from Wave 1:

1. Files are **SPSS `.sav`**, not CSV — loaded with `pandas.read_spss` (via `pyreadstat`).
2. SPSS value labels are **auto-decoded** — e.g. `saq01` (region) already reads `"Tigray"`, `"Amhara"`, ... instead of a numeric code, and `crop_code` already reads `"MAIZE"`, `"TEFF"`, ... More free human-readable columns than Wave 1, but not everything is labeled (loss-reason and disposition-channel codes are still raw numbers).

| File | Assumed role | Verified role |
|---|---|---|
| `sect1_hh_w2` | household + region identifier | Confirmed: person roster (`hh_s1q00`=person number, `hh_s1q02`=relationship to head **already text**, `hh_s1q03`=sex **already text**, `hh_s1q04_a`=age) + region. Same caveat as Wave 1 — region/geo codes exist in every file below. |
| `sect1_ph_w2` | post-harvest crop list | **Not crop data**, same as Wave 1 — a PH-module respondent roster, no crop column. |
| `sectnr_ph_w2` | post-harvest non-roster details | Same as Wave 1 — non-roster holder IDs, no crop/loss info. |
| `sect9_ph_w2` | crop disposition (sold/stored/lost) | **Role changed from Wave 1.** Now general crop/parcel characteristics (`ph_s9q01` = cropping pattern, "PURE STAND"/"MIXED STAND", already text) — not storage or disposition. |
| *(new)* `sect9a_ph_w2` | — | **The real storage-detail file** — direct analog of Wave 1's `sect9_ph_w1` (same quantity/date-pair structure). Wave 2 split Wave 1's single section into two; using `sect9a` for the storage role instead of `sect9`. |
| `sect10_ph_w2` | crop loss quantities + reasons | **Confirmed**, same triplet structure as Wave 1. Reason codes still raw numbers. |
| `sect11_ph_w2` | tree/root crops post-harvest | Same as Wave 1: field crops, and `ph_s11q22_a`-`g` is the confirmed disposition % breakdown. **Not** tree/root crops. |
| `sect12_ph_w2` | storage type used | Same as Wave 1: garden/perennial crops with an analogous disposition % structure. **Not** storage type. |

**Column names match Wave 1's `crop_loss_master` schema exactly** (reindexed to it at the end), so this and every later wave's output stays directly concatenable. Two data-quality fixes are needed here that Wave 1 didn't need:

- **Categorical groupby blow-up:** `pandas.read_spss` returns every labeled column (crop_code, region_code, ...) as a `Categorical`. Grouping by a `Categorical` with `dropna=False` enumerates *every* category combination, including unused ones — this silently exploded memory during development. Fixed by casting all categorical columns to plain strings before grouping.
- **Blank `household_id` rows:** a genuine data-quality issue in the raw SPSS export — a small number of rows (257 in `sect10`, 119 in `sect11`, 83 in `sect12`, 5,248 in `sect1_hh`) have a blank (not `NaN`) `household_id`. Left in, `dropna=False` groups them into one bogus mega-household. Dropped before aggregating.

```python
DATA_DIR_W2 = Path("../data/raw/ETH_2013_ESS_v03_M_SPSS")
sorted(p.name for p in DATA_DIR_W2.glob("*.sav"))[:10]
```

```python
# The 86-column schema every later wave gets reindexed to, captured from
# Wave 1's finished output (minus survey_year, added separately per wave).
BASE_SCHEMA = [c for c in crop_loss_master.columns if c != "survey_year"]

GEO_RENAME_W2 = {
    "saq01": "region_code", "saq02": "zone_code", "saq03": "woreda_code",
    "saq04": "town_code", "saq05": "kebele_code", "saq06": "ea_code_local",
    "rural": "is_rural", "pw2": "sampling_weight",
    "household_id": "household_id", "holder_id": "holder_id", "ea_id": "ea_id",
    "crop_name": "crop_name", "crop_code": "crop_code",
}

# sect9a (storage): q02/q04 verified as (day, month) date pairs -- unused
# here since Wave 1's slots are quantity/unit, not dates -- the two clean
# quantity pairs (q03, q05) map to storage_qty1/2.
SECT9A_RENAME_W2 = {
    **GEO_RENAME_W2, "parcel_id": "parcel_id", "field_id": "field_id",
    "ph_s9q01": "storage_flag_unconfirmed",
    "ph_s9q03_a": "storage_qty1_unit_unconfirmed", "ph_s9q03_b": "storage_qty1_unconfirmed",
    "ph_s9q05_a": "storage_qty2_unit_unconfirmed", "ph_s9q05_b": "storage_qty2_unconfirmed",
}

# sect10 (loss): CONFIRMED identical triplet/lettering structure to Wave 1.
SECT10_RENAME_W2 = {
    **GEO_RENAME_W2, "parcel_id": "parcel_id", "field_id": "field_id",
    "ph_s10q01_a": "loss_reason1_occurred", "ph_s10q01_b": "loss_reason1_unit", "ph_s10q01_c": "loss_reason1_qty",
    "ph_s10q01_d": "loss_reason2_occurred", "ph_s10q01_e": "loss_reason2_unit", "ph_s10q01_f": "loss_reason2_qty",
    "ph_s10q01_g": "loss_reason3_occurred", "ph_s10q01_h": "loss_reason3_unit", "ph_s10q01_i": "loss_reason3_qty",
}
for c in [f"ph_s10q02_{s}" for s in "abcdefghijklmnopqrstuvwxyz"] + [f"ph_s10q02_{s}a" for s in "klmnop"]:
    SECT10_RENAME_W2[c] = f"loss_detail_{c.split('_')[-1]}_unconfirmed"
for s in "abcdef":
    SECT10_RENAME_W2[f"ph_s10q03_{s}"] = f"loss_extra_{s}_unconfirmed"


def disposition_rename_w2(question_prefix, letters):
    d = {**GEO_RENAME_W2}
    for i, l in enumerate(letters[:-1], start=1):
        d[f"{question_prefix}_{l}"] = f"disposition_share{i}_pct_unconfirmed"
    d[f"{question_prefix}_{letters[-1]}"] = "disposition_total_check_pct"
    return d


SECT11_RENAME_W2 = {**GEO_RENAME_W2, **disposition_rename_w2("ph_s11q22", "abcdefg")}
SECT12_RENAME_W2 = {**GEO_RENAME_W2, **disposition_rename_w2("ph_s12q19", "abcefgh")}  # skips letter 'd'

SECT1_HH_RENAME_W2 = {
    **GEO_RENAME_W2, "individual_id": "individual_id",
    "hh_s1q00": "person_number", "hh_s1q03": "sex", "hh_s1q04_a": "age",
}

df9a_w2 = pd.read_spss(DATA_DIR_W2 / "sect9a_ph_w2.sav").rename(columns=SECT9A_RENAME_W2)
df10_w2 = pd.read_spss(DATA_DIR_W2 / "sect10_ph_w2.sav").rename(columns=SECT10_RENAME_W2)
df11_w2 = pd.read_spss(DATA_DIR_W2 / "sect11_ph_w2.sav").rename(columns=SECT11_RENAME_W2)
df12_w2 = pd.read_spss(DATA_DIR_W2 / "sect12_ph_w2.sav").rename(columns=SECT12_RENAME_W2)
df1_hh_w2 = pd.read_spss(DATA_DIR_W2 / "sect1_hh_w2.sav").rename(columns=SECT1_HH_RENAME_W2)

# Blank (not NaN) household_id rows -- see Part 3 header. Drop before anything
# else touches these frames.
for _name in ["df9a_w2", "df10_w2", "df11_w2", "df12_w2", "df1_hh_w2"]:
    _frame = globals()[_name]
    _before = len(_frame)
    _frame = _frame[_frame["household_id"].astype(str).str.strip() != ""]
    _dropped = _before - len(_frame)
    if _dropped:
        print(f"{_name}: dropped {_dropped} rows with blank household_id")
    globals()[_name] = _frame

# sect9a/sect10 only carry crop_code (already decoded to crop-name text);
# sect11/sect12 carry both crop_name and crop_code (identical text). Backfill
# whichever is missing so all four have both, then de-dup the resulting
# duplicate columns on sect11/sect12.
for _name in ["df9a_w2", "df10_w2", "df11_w2", "df12_w2"]:
    _frame = globals()[_name]
    if "crop_code" not in _frame.columns and "crop_name" in _frame.columns:
        _frame["crop_code"] = _frame["crop_name"]
    if "crop_name" not in _frame.columns and "crop_code" in _frame.columns:
        _frame["crop_name"] = _frame["crop_code"]
    if _frame.columns.duplicated().any():
        globals()[_name] = _frame.loc[:, ~_frame.columns.duplicated()]

# Categorical groupby blow-up fix -- see Part 3 header. Cast every
# pandas.read_spss Categorical column to plain string before any groupby.
for _name in ["df9a_w2", "df10_w2", "df11_w2", "df12_w2", "df1_hh_w2"]:
    _frame = globals()[_name]
    for c in _frame.select_dtypes("category").columns:
        _frame[c] = _frame[c].astype(str).replace("nan", pd.NA)

# region_code comes through SPSS-decoded as region-NAME text (e.g. "Afar")
# for this wave only; map back to the same numeric code every other wave
# uses (confirmed via sect1_hh_w2's saq01 value labels), so region_code is
# a consistent int across all 5 waves' output.
REGION_NAME_TO_CODE_W2 = {
    "Tigray": 1, "Afar": 2, "Amhara": 3, "Oromia": 4, "Somalie": 5,
    "Benshagul Gumuz": 6, "SNNP": 7, "Gambelia": 12, "Harari": 13,
    "Addis Ababa": 14, "Diredwa": 15,
}
for _name in ["df9a_w2", "df10_w2", "df11_w2", "df12_w2", "df1_hh_w2"]:
    _frame = globals()[_name]
    if "region_code" in _frame.columns:
        _frame["region_code"] = _frame["region_code"].map(REGION_NAME_TO_CODE_W2).astype("Int64")

# is_rural: Wave 1 was already 0/1; Wave 2's `rural` decodes to text
# ("Rural"/"Small town (urban)"/"Large town (urban)") -- collapse to the
# same binary semantics as Wave 1.
for _name in ["df9a_w2", "df10_w2", "df11_w2", "df12_w2", "df1_hh_w2"]:
    _frame = globals()[_name]
    if "is_rural" in _frame.columns:
        _frame["is_rural"] = (_frame["is_rural"].astype(str) == "Rural").astype(int)

df9a_w2.shape, df10_w2.shape, df11_w2.shape, df12_w2.shape, df1_hh_w2.shape
```

## Part 5 — Waves 4 & 5 (2018-19, 2021-22)

Waves 4-5 use a **substantially redesigned questionnaire**: no more `ph_` prefix (`s10q01a` instead of `ph_s10q01_a`), and crop identity is no longer a separate `crop_code`/`crop_name` pair — it's baked inline into a single text field as `"N. CROPNAME"` (e.g. `s9q00b` = `"6. SORGHUM"`), which gets split into numeric code + name here.

- **Storage (`sect9`) and loss (`sect10`) are solid.** `sect10`'s `s10q01a`-`i` has the exact same confirmed 3×(occurred, unit, qty) triplet structure as every earlier wave. Crop identity comes through cleanly as English text.
- **Disposition (`sect11`) had to be dropped for these two waves.** It only carries `harvestedcrop_id` (a small integer) and its own crop-name field, which is mostly garbled Amharic-script text with a few inconsistent romanized entries. `harvestedcrop_id` doesn't resolve back to `sect9`'s crop list either — `sect9`'s crop numbering resets per field, so the join is ambiguous for the majority of rows. Decided with the user to leave `disposition_share*`/`crop_domain` empty for 2018 and 2021 rather than attach them to unreliable crop names.
- Several Wave 1-3 `storage_*_unconfirmed` slots don't have an obvious analog in the redesigned `sect9` and are left empty for these two waves — only the flag (`s9q04`) and one clean quantity field (`s9q05a`) map cleanly.
- `household_id` representation is inconsistent across files within these waves too (int in some, zero-padded string in others) — normalized through a numeric round-trip before any merge, the same fix needed during development (a naive string cast would leave a leading-zero mismatch that silently fails every join).

```python
import re

LABEL_RE = re.compile(r"^\s*(\d+)\.\s*(.+?)\s*$")


def split_code_label(series):
    """'6. SORGHUM' -> (6, 'SORGHUM'); leaves non-matching values as (NaN, original)."""
    codes, labels = [], []
    for v in series:
        if pd.isna(v):
            codes.append(pd.NA)
            labels.append(pd.NA)
            continue
        m = LABEL_RE.match(str(v))
        if m:
            codes.append(int(m.group(1)))
            labels.append(m.group(2).upper())
        else:
            codes.append(pd.NA)
            labels.append(str(v).upper())
    return pd.Series(codes, index=series.index), pd.Series(labels, index=series.index)


def build_wave_redesigned(wave_suffix, data_dir, survey_year):
    """Shared pipeline for Waves 4 (2018) and 5 (2021)."""
    print(f"\n{'='*20} Wave {survey_year} ({wave_suffix}) {'='*20}")

    geo_rename = {
        "saq01": "region_code", "saq02": "zone_code", "saq03": "woreda_code",
        "saq04": "town_code", "saq05": "kebele_code", "saq06": "ea_code_local",
        "ea_id": "ea_id", "saq14": "is_rural", f"pw_{wave_suffix}": "sampling_weight",
        "household_id": "household_id", "holder_id": "holder_id",
    }
    sect9_rename = {
        **geo_rename, "parcel_id": "parcel_id", "field_id": "field_id",
        "s9q04": "storage_flag_unconfirmed",
        "s9q05a": "storage_qty1_unconfirmed",
    }
    sect10_rename = {
        **geo_rename, "parcel_id": "parcel_id", "field_id": "field_id",
        "s10q01a": "loss_reason1_occurred", "s10q01b": "loss_reason1_unit", "s10q01c": "loss_reason1_qty",
        "s10q01d": "loss_reason2_occurred", "s10q01e": "loss_reason2_unit", "s10q01f": "loss_reason2_qty",
        "s10q01g": "loss_reason3_occurred", "s10q01h": "loss_reason3_unit", "s10q01i": "loss_reason3_qty",
    }
    for l in "abcdefghijklmnop":
        sect10_rename[f"s10q02{l}"] = f"loss_detail_{l}_unconfirmed"
    for l in "abcdef":
        sect10_rename[f"s10q03{l}"] = f"loss_extra_{l}_unconfirmed"

    df9 = pd.read_csv(data_dir / f"sect9_ph_{wave_suffix}.csv", low_memory=False).rename(columns=sect9_rename)
    df10 = pd.read_csv(data_dir / f"sect10_ph_{wave_suffix}.csv", low_memory=False).rename(columns=sect10_rename)
    df1_hh = pd.read_csv(data_dir / f"sect1_hh_{wave_suffix}.csv", low_memory=False).rename(columns=geo_rename)

    # household_id dtype varies file-to-file (int64 in some files, a
    # zero-padded string in others). A numeric round-trip (pd.to_numeric ->
    # Int64) silently downcasts these ~17-digit IDs to float64 first and
    # loses precision (float64 only holds integers exactly up to 2**53),
    # corrupting most IDs. Normalize as pure strings instead: strip
    # whitespace, drop a trailing ".0" left by any prior float cast, then
    # strip leading zeros so padded and unpadded forms match.
    def normalize_household_id(s):
        return s.astype(str).str.strip().str.replace(r"\.0$", "", regex=True).str.lstrip("0")

    for frame in (df9, df10, df1_hh):
        frame["household_id"] = normalize_household_id(frame["household_id"])

    for name, frame in [("sect9", df9), ("sect10", df10), ("sect1_hh", df1_hh)]:
        before = len(frame)
        frame = frame[frame["household_id"].notna()]
        dropped = before - len(frame)
        if dropped:
            print(f"{name}: dropped {dropped} rows with missing household_id")
        if name == "sect9":
            df9 = frame
        elif name == "sect10":
            df10 = frame
        else:
            df1_hh = frame

    df9["crop_code"], df9["crop_name"] = split_code_label(df9["s9q00b"])
    df10["crop_code"], df10["crop_name"] = split_code_label(df10["s9q00b"])

    for frame in (df9, df10, df1_hh):
        frame["region_code"], _ = split_code_label(frame["region_code"])
        _, rural_label = split_code_label(frame["is_rural"])
        frame["is_rural"] = (rural_label == "RURAL").astype("Int64")

    grain = ["household_id", "crop_code", "crop_name"]
    non_additive = {
        "holder_id", "ea_id", "parcel_id", "field_id",
        "region_code", "zone_code", "woreda_code", "town_code", "kebele_code",
        "ea_code_local", "is_rural", "sampling_weight",
    }

    def agg_to_crop_grain(df):
        numeric_cols = df.select_dtypes("number").columns.difference(grain + list(non_additive))
        return df.groupby(grain, dropna=False)[list(numeric_cols)].sum(min_count=1).reset_index()

    storage_agg = agg_to_crop_grain(df9)
    loss_agg = agg_to_crop_grain(df10)
    loss_agg["total_loss_qty"] = loss_agg[["loss_reason1_qty", "loss_reason2_qty", "loss_reason3_qty"]].sum(axis=1, min_count=1)

    household_summary = (
        df1_hh.groupby("household_id")
        .agg(
            household_size=("individual_id", "count"),
            region_code=("region_code", "first"),
            zone_code=("zone_code", "first"),
            woreda_code=("woreda_code", "first"),
            is_rural=("is_rural", "first"),
        )
        .reset_index()
    )

    crop_loss_master = (
        storage_agg.merge(loss_agg, on=grain, how="outer", suffixes=("", "_loss"))
        .merge(household_summary, on="household_id", how="left")
    )
    # No disposition data for these waves (see Part 5 header).
    crop_loss_master["crop_domain"] = pd.NA

    dupes = [c for c in crop_loss_master.columns if c.endswith("_x") or c.endswith("_y")]
    assert not dupes, f"duplicate-suffixed columns: {dupes}"

    crop_loss_master = crop_loss_master.reindex(columns=BASE_SCHEMA)
    assert list(crop_loss_master.columns) == BASE_SCHEMA
    crop_loss_master["survey_year"] = survey_year

    out_path = Path(f"../data/processed/crop_loss_master_{survey_year}.csv")
    crop_loss_master.to_csv(out_path, index=False)
    print(f"saved {crop_loss_master.shape} -> {out_path}")
    print("nan household_id:", crop_loss_master["household_id"].isna().sum())
    print("rows with total_loss_qty > 0:", (crop_loss_master["total_loss_qty"] > 0).sum())
    print("household_size describe:")
    print(crop_loss_master["household_size"].describe())
    return crop_loss_master


crop_loss_master_w4 = build_wave_redesigned("w4", Path("../data/raw/ETH_2018_ESS_v04_M_CSV"), 2018)
crop_loss_master_w5 = build_wave_redesigned("w5", Path("../data/raw/ETH_2021_ESPS-W5_v02_M_CSV"), 2021)
```

## Part 8 — CHIRPS rainfall: explore & select columns for a future merge

Source: `data/raw/CHIRPS/eth-rainfall-subnat-full.csv` (WFP/CHIRPS dekadal rainfall
indicators for Ethiopia, downloaded from data.humdata.org). Raw structure:

- `date`: dekadal (10-day period) observations -- the 1st, 11th, and 21st of every
  month, 1981-present.
- `adm_level` / `adm_id` / `PCODE`: administrative unit. `adm_level=1` is region
  (11 regions, `PCODE` like `ET01`..`ET15`); `adm_level=2` is zone (74 zones,
  `PCODE` like `ET0101`).
- `rfh` / `rfh_avg`: rainfall for the current dekad (mm) and its long-term
  average for that same dekad.
- `r1h` / `r1h_avg`, `r3h` / `r3h_avg`: rolling 1-month and 3-month rainfall
  totals and their long-term averages.
- `rfq` / `r1q` / `r3q`: the current dekad/1-month/3-month rainfall as a
  percentage of its long-term average (i.e. a rainfall anomaly / drought-flood
  indicator).
- `n_pixels`, `version` (`final`/`prelim`): metadata -- pixel count behind the
  spatial average, and whether the estimate is final or a preliminary value
  for the most recent month.

**Region-code crosswalk confirmed**: `sect1_hh_w2`'s SPSS value labels for
`saq01` (region_code) decode exactly as `1=Tigray, 2=Afar, 3=Amhara, 4=Oromia,
5=Somalie, 6=Benshagul Gumuz, 7=SNNP, 12=Gambelia, 13=Harari, 14=Addis Ababa,
15=Diredwa` -- i.e. the LSMS numeric `region_code` used in
`crop_loss_master_all` **is** the numeric suffix of CHIRPS's region `PCODE`
(region_code 14 <-> `ET14`, etc.). No external boundary lookup needed.

This step keeps only region-level rows (`adm_level == 1` -- LSMS's `zone_code`
is a per-region local sequence, not nationally unique, so it can't be matched
to CHIRPS zone PCODEs without more work) and only `final` rows (the CHIRPS
years relevant here, 2011-2021, are all final; `prelim` only covers the most
recent month of the download). Still at dekadal grain -- aggregating to
annual/seasonal rainfall features and actually merging onto
`crop_loss_master_all` is a follow-up step, not done here.

```python
rain_raw = pd.read_csv("../data/raw/CHIRPS/eth-rainfall-subnat-full.csv", low_memory=False)
print("raw shape:", rain_raw.shape)
print(rain_raw.dtypes)

RAINFALL_RENAME = {
    "date": "date",
    "PCODE": "admin_pcode",
    "rfh": "rainfall_dekad_mm",
    "rfh_avg": "rainfall_dekad_avg_mm",
    "r1h": "rainfall_1month_mm",
    "r1h_avg": "rainfall_1month_avg_mm",
    "r3h": "rainfall_3month_mm",
    "r3h_avg": "rainfall_3month_avg_mm",
    "rfq": "rainfall_dekad_pct_of_avg",
    "r1q": "rainfall_1month_pct_of_avg",
    "r3q": "rainfall_3month_pct_of_avg",
    "version": "data_version",
}

rainfall_region = (
    rain_raw[rain_raw["adm_level"] == 1]
    .query("version == \'final\'")
    [list(RAINFALL_RENAME)]
    .rename(columns=RAINFALL_RENAME)
    .copy()
)

rainfall_region["region_code"] = rainfall_region["admin_pcode"].str.replace("ET", "", regex=False).astype(int)
rainfall_region["date"] = pd.to_datetime(rainfall_region["date"])
rainfall_region["year"] = rainfall_region["date"].dt.year
rainfall_region["month"] = rainfall_region["date"].dt.month

rainfall_region = rainfall_region[
    ["date", "year", "month", "region_code", "admin_pcode",
     "rainfall_dekad_mm", "rainfall_dekad_avg_mm", "rainfall_dekad_pct_of_avg",
     "rainfall_1month_mm", "rainfall_1month_avg_mm", "rainfall_1month_pct_of_avg",
     "rainfall_3month_mm", "rainfall_3month_avg_mm", "rainfall_3month_pct_of_avg",
     "data_version"]
]

out_path = Path("../data/processed/rainfall_region_dekadal.csv")
rainfall_region.to_csv(out_path, index=False)
print(f"\nsaved: {rainfall_region.shape} -> {out_path}")
print("\nregion_codes:", sorted(rainfall_region["region_code"].unique()))
print("year range:", rainfall_region["year"].min(), "-", rainfall_region["year"].max())
rainfall_region.head()
```

## Part 9 — Aggregate rainfall to season/year and merge into `crop_loss_master_all`

`crop_loss_master_all` has one row per household+crop **per survey wave** (no
exact interview date), so the dekadal rainfall from Part 8 has to be
aggregated up before it can be joined -- by `region_code` and by calendar
year, matched to `survey_year`. Ethiopia has two rainy seasons that matter for
crop outcomes:

- **Belg** (Feb-May): short rains, secondary season in most regions.
- **Meher** (Jun-Sep): main rains, ~90%+ of national production; this is the
  season LSMS's post-harvest questions are mostly asking about.

For each `(region_code, year)` we compute total rainfall (mm) and rainfall as
a % of the long-term average, for Belg, Meher, and the full year, then merge
onto `crop_loss_master_all` on `(region_code, survey_year == year)`.
**Assumption, not confirmed against survey field-dates**: this matches each
wave's rainfall to the same calendar year as `survey_year` (e.g. 2018 crop
loss rows get 2018's Belg/Meher rainfall). LSMS post-harvest interviews are
fielded after the Meher harvest (roughly Sept-Jan), so this should be the
right growing season for most rows, but a household interviewed early in a
wave could in principle be reporting on the *previous* year's Meher. Treat it
the same way as the project's other `_unconfirmed` fields if that matters for
the model.

**Design note on `region_code` vs `region_name` (and the same question for
any other coded field)**: both are kept, on purpose, rather than picking one.
`region_code` is the join/model key everywhere in this pipeline -- it's
stable and unambiguous, whereas the *text* spellings turned out to vary by
source even within this one project (Wave 2's raw SPSS labels alone had
"Somalie"/"Diredwa"/"Benshagul Gumuz" for what a person would call
"Somali"/"Dire Dawa"/"Benishangul-Gumuz" -- see the Wave 2 fix above). For
anything the mobile app actually displays, use `region_name` (or a cleaned
label like it) -- never show a raw numeric code or an internal machine key to
an end user. The same split applies to the rainfall columns here: the model
can consume `rainfall_meher_pct_of_avg` directly, but a UI should show
something like "Meher rainfall: 78% of normal" rather than the column name.
Keep the code as the source of truth for joins/training, and maintain the
human label as a lookup applied only at the display layer -- that way a
relabeling (or a translation) never risks silently breaking a merge.

```python
rainfall_dekadal = pd.read_csv("../data/processed/rainfall_region_dekadal.csv")

SEASON_MONTHS = {
    "belg": range(2, 6),    # Feb-May
    "meher": range(6, 10),  # Jun-Sep
}

def season_agg(df, months, label):
    sub = df[df["month"].isin(months)]
    out = (
        sub.groupby(["region_code", "year"])
        .agg(_total_mm=("rainfall_dekad_mm", "sum"), _avg_mm=("rainfall_dekad_avg_mm", "sum"))
        .reset_index()
    )
    out[f"rainfall_{label}_mm"] = out["_total_mm"]
    out[f"rainfall_{label}_pct_of_avg"] = (out["_total_mm"] / out["_avg_mm"] * 100).round(1)
    return out[["region_code", "year", f"rainfall_{label}_mm", f"rainfall_{label}_pct_of_avg"]]

belg_agg = season_agg(rainfall_dekadal, SEASON_MONTHS["belg"], "belg")
meher_agg = season_agg(rainfall_dekadal, SEASON_MONTHS["meher"], "meher")

annual_agg = (
    rainfall_dekadal.groupby(["region_code", "year"])
    .agg(rainfall_annual_mm=("rainfall_dekad_mm", "sum"), _avg_mm=("rainfall_dekad_avg_mm", "sum"))
    .reset_index()
)
annual_agg["rainfall_annual_pct_of_avg"] = (annual_agg["rainfall_annual_mm"] / annual_agg["_avg_mm"] * 100).round(1)
annual_agg = annual_agg[["region_code", "year", "rainfall_annual_mm", "rainfall_annual_pct_of_avg"]]

rainfall_seasonal = (
    belg_agg.merge(meher_agg, on=["region_code", "year"], how="outer")
    .merge(annual_agg, on=["region_code", "year"], how="outer")
)
rainfall_seasonal["region_code"] = rainfall_seasonal["region_code"].astype("Int64")

crop_loss_master_all = pd.read_csv("../data/processed/crop_loss_master_all.csv", low_memory=False)
crop_loss_master_all["region_code"] = pd.to_numeric(crop_loss_master_all["region_code"], errors="coerce").astype("Int64")

rain_cols = [c for c in rainfall_seasonal.columns if c not in ("region_code", "year")]
crop_loss_master_all = crop_loss_master_all.drop(columns=[c for c in rain_cols if c in crop_loss_master_all.columns])
crop_loss_master_all = crop_loss_master_all.merge(
    rainfall_seasonal, left_on=["region_code", "survey_year"], right_on=["region_code", "year"], how="left"
).drop(columns="year")

crop_loss_master_all.to_csv("../data/processed/crop_loss_master_all.csv", index=False)

print("final shape:", crop_loss_master_all.shape)
print("\nrainfall column null rate by survey_year:")
print(crop_loss_master_all.groupby("survey_year")[rain_cols].apply(lambda g: g.isna().mean()))
print("\nmean Meher rainfall (% of average) by survey_year:")
print(crop_loss_master_all.groupby("survey_year")["rainfall_meher_pct_of_avg"].mean())
```
