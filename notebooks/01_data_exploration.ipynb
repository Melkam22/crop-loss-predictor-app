## Part 2 — Verified file roles & crop-level merge

Checking actual column content (crop lists, value ranges, percentage-sum patterns) against the originally assumed file roles turned up several corrections:

| File | Assumed role | Verified role |
|---|---|---|
| `sect1_hh_w1` | household + region identifier | Household **person roster** (age, sex, ...) + region codes. Region codes (`saq01`-`06`) actually exist identically in every PH file below, not just this one. |
| `sect1_ph_w1` | post-harvest crop list | **Not crop data** — no `crop_name` column at all. It's a person roster for the PH module respondent. |
| `sect_nr_ph_w1` | post-harvest non-roster details | Confirmed-ish: holder IDs for people *not* in the household roster. No crop/loss info. |
| `sect9_ph_w1` | crop disposition (sold/stored/lost) | **Storage** detail for field crops (cereals/pulses/oilseeds) — quantity/unit pairs, not disposition. |
| `sect10_ph_w1` | crop loss quantities + reasons | **Confirmed.** 9 lettered columns are 3 (occurred-flag, unit, quantity) triplets = 3 loss-reason slots per crop. |
| `sect11_ph_w1` | tree/root crops post-harvest | Field crops (same scope as sect9). Contains `q22_a`-`g`, percentages summing to ~100 — this is the **disposition breakdown** (sold/consumed/stored/lost shares). |
| `sect12_ph_w1` | storage type used | Garden/perennial crops (fruit, veg, spice). Same %-share disposition structure as sect11, for the garden-crop domain. This is the "tree/root/garden crop" file, not "storage type". |

**Grain mismatch:** `sect9`/`sect10` are one row per household-parcel-field-crop; `sect11`/`sect12` are one row per household-crop (already aggregated across parcels). `sect1_hh_w1` is one row per person. Merging all of these directly on `household_id` would fan out crop rows across household members. Plan:

1. Load only the 5 files that carry crop/loss/person signal: `sect9`, `sect10`, `sect11`, `sect12`, `sect1_hh`.
2. Rename columns to human-readable names — confirmed meanings get real names; everything still ambiguous keeps an honest generic name (`_unconfirmed`) rather than a guessed one, since inventing labels here would be actively misleading without the official LSMS codebook.
3. Aggregate `sect9` and `sect10` up to household+crop grain (sum across parcels/fields).
4. Concatenate `sect11` (field crops) + `sect12` (garden crops) into one disposition table.
5. Outer-merge storage + loss + disposition on `household_id` + `crop_code` + `crop_name`.
6. Attach household-level info (household size, rural flag, region codes) from `sect1_hh_w1`, aggregated to one row per household.

```python
# Geographic hierarchy (saq01-06) follows the standard published ERSS/LSMS-ISA
# Ethiopia admin coding (region > zone > woreda > town > kebele > ea). Kept as
# *_code, not decoded to names, since we don't have the region/zone code lookup.
GEO_RENAME = {
    "saq01": "region_code",
    "saq02": "zone_code",
    "saq03": "woreda_code",
    "saq04": "town_code",
    "saq05": "kebele_code",
    "saq06": "ea_code_local",
    "ea_id": "ea_id",
    "rural": "is_rural",
    "pw": "sampling_weight",
    "household_id": "household_id",
    "holder_id": "holder_id",
    "crop_name": "crop_name",
    "crop_code": "crop_code",
}

# --- sect9: storage detail (field crops) ---
# q01 is binary-like (stored y/n). q02-q05 and q08/q12/q13 are (quantity, unit)
# pairs but WHICH storage attribute each pair measures isn't decodable without
# the codebook, so they keep generic sequential names rather than guessed ones.
SECT9_RENAME = {
    **GEO_RENAME,
    "parcel_id": "parcel_id",
    "field_id": "field_id",
    "ph_s9q01": "storage_flag_unconfirmed",
    "ph_s9q02_a": "storage_qty1_unconfirmed", "ph_s9q02_b": "storage_qty1_unit_unconfirmed",
    "ph_s9q03_a": "storage_qty2_unconfirmed", "ph_s9q03_b": "storage_qty2_unit_unconfirmed",
    "ph_s9q04_a": "storage_qty3_unconfirmed", "ph_s9q04_b": "storage_qty3_unit_unconfirmed",
    "ph_s9q05_a": "storage_qty4_unconfirmed", "ph_s9q05_b": "storage_qty4_unit_unconfirmed",
    "ph_s9q06": "storage_flag2_unconfirmed",
    "ph_s9q07": "storage_flag3_unconfirmed",
    "ph_s9q08_a": "storage_qty5_unconfirmed", "ph_s9q08_b": "storage_qty5_unit_unconfirmed",
    "ph_s9q09": "storage_flag4_unconfirmed",
    "ph_s9q10": "storage_detail1_unconfirmed",
    "ph_s9q11": "storage_detail2_unconfirmed",
    "ph_s9q12_a": "storage_qty6_unconfirmed", "ph_s9q12_b": "storage_qty6_unit_unconfirmed",
    "ph_s9q13_a": "storage_qty7_unconfirmed", "ph_s9q13_b": "storage_qty7_unit_unconfirmed",
}

# --- sect10: loss reasons + quantities (all crop domains) --- CONFIRMED structure:
# 3 reason "slots", each a (occurred, unit, quantity) triplet. Which specific
# reason (pest/rain/theft/...) each slot represents is not decodable without
# the codebook, so slots are numbered, not named by cause.
SECT10_RENAME = {
    **GEO_RENAME,
    "parcel_id": "parcel_id",
    "field_id": "field_id",
    "crop_id": "crop_instance_id",
    "ph_s10q01_a": "loss_reason1_occurred", "ph_s10q01_b": "loss_reason1_unit", "ph_s10q01_c": "loss_reason1_qty",
    "ph_s10q01_d": "loss_reason2_occurred", "ph_s10q01_e": "loss_reason2_unit", "ph_s10q01_f": "loss_reason2_qty",
    "ph_s10q01_g": "loss_reason3_occurred", "ph_s10q01_h": "loss_reason3_unit", "ph_s10q01_i": "loss_reason3_qty",
}
# q02 (33 cols) and q03 (6 cols) blocks: pattern not confidently decoded, kept generic
for c in [f"ph_s10q02_{s}" for s in "abcdefghijklmnopqrstuvwxyz"] + [f"ph_s10q02_{s}a" for s in "klmnop"]:
    SECT10_RENAME[c] = f"loss_detail_{c.split('_')[-1]}_unconfirmed"
for s in "abcdef":
    SECT10_RENAME[f"ph_s10q03_{s}"] = f"loss_extra_{s}_unconfirmed"

# --- sect11 / sect12: disposition share breakdown ---
# q22 (sect11) / q19 (sect12), columns a-f are percentage shares of the crop's
# disposition that sum to ~100; the final letter (g) is a total/check column.
# Exact category per share (sold/consumed/seed/...) is not decodable without
# the codebook.
def disposition_rename(question_prefix):
    d = {**GEO_RENAME}
    letters = "abcdefg"
    for i, l in enumerate(letters[:6], start=1):
        d[f"{question_prefix}_{l}"] = f"disposition_share{i}_pct_unconfirmed"
    d[f"{question_prefix}_g"] = "disposition_total_check_pct"
    return d

SECT11_RENAME = {**GEO_RENAME, **disposition_rename("ph_s11q22")}
SECT12_RENAME = {**GEO_RENAME, **disposition_rename("ph_s12q19")}

# --- sect1_hh: household roster --- CONFIRMED: q00=person number, q03=sex, q04_a=age
SECT1_HH_RENAME = {
    **GEO_RENAME,
    "individual_id": "individual_id",
    "hh_s1q00": "person_number",
    "hh_s1q03": "sex",
    "hh_s1q04_a": "age",
}

df9 = pd.read_csv(DATA_DIR / "sect9_ph_w1.csv").rename(columns=SECT9_RENAME)
df10 = pd.read_csv(DATA_DIR / "sect10_ph_w1.csv").rename(columns=SECT10_RENAME)
df11 = pd.read_csv(DATA_DIR / "sect11_ph_w1.csv").rename(columns=SECT11_RENAME)
df12 = pd.read_csv(DATA_DIR / "sect12_ph_w1.csv").rename(columns=SECT12_RENAME)
df1_hh = pd.read_csv(DATA_DIR / "sect1_hh_w1.csv").rename(columns=SECT1_HH_RENAME)

df9.shape, df10.shape, df11.shape, df12.shape, df1_hh.shape
```

### Aggregate parcel/field-level files up to household + crop grain

`sect9` (storage) and `sect10` (loss) have one row per parcel/field within a crop; `sect11`/`sect12` (disposition) are already one row per household-crop. Sum the numeric quantity columns across parcels/fields so every table ends up at the same `household_id` + `crop_code` grain.

```python
GRAIN = ["household_id", "crop_code", "crop_name"]

# Identifier/geo/weight columns are numeric but must NOT be summed across
# parcels/fields (they're constant per household, not additive quantities).
NON_ADDITIVE_COLS = {
    "holder_id", "ea_id", "parcel_id", "field_id", "crop_instance_id",
    "region_code", "zone_code", "woreda_code", "town_code", "kebele_code",
    "ea_code_local", "is_rural", "sampling_weight",
}

def agg_to_crop_grain(df, grain=GRAIN):
    numeric_cols = df.select_dtypes("number").columns.difference(grain + list(NON_ADDITIVE_COLS))
    agg = df.groupby(grain, dropna=False)[list(numeric_cols)].sum(min_count=1)
    return agg.reset_index()

storage_agg = agg_to_crop_grain(df9)
loss_agg = agg_to_crop_grain(df10)

# Engineered convenience column: total quantity lost across the 3 reason slots.
# Safe to sum because all three are the same "qty" field by construction (verified
# structure), regardless of which specific reason each slot represents.
loss_agg["total_loss_qty"] = loss_agg[
    ["loss_reason1_qty", "loss_reason2_qty", "loss_reason3_qty"]
].sum(axis=1, min_count=1)

storage_agg.shape, loss_agg.shape
```

### Combine disposition tables (field crops + garden crops)

```python
df11["crop_domain"] = "field_crop"
df12["crop_domain"] = "garden_crop"

disposition_cols = GRAIN + ["crop_domain"] + [
    c for c in df11.columns if c.startswith("disposition_")
]
disposition_all = pd.concat(
    [df11[disposition_cols], df12[disposition_cols]],
    ignore_index=True,
)

# sect11/sect12 are already one row per household-crop, but guard against
# duplicate interviews by keeping the first record per household+crop.
disposition_all = disposition_all.drop_duplicates(subset=GRAIN, keep="first")

disposition_all.shape
```

### Household-level summary (from the person roster)

`sect1_hh_w1` is per-person; collapse it to one row per household (size + region/rural flags) before attaching it to the crop-level table.

```python
household_summary = (
    df1_hh.groupby("household_id")
    .agg(
        household_size=("person_number", "count"),
        region_code=("region_code", "first"),
        zone_code=("zone_code", "first"),
        woreda_code=("woreda_code", "first"),
        is_rural=("is_rural", "first"),
    )
    .reset_index()
)

household_summary.shape
```

### Final merge: one row per household + crop

```python
crop_loss_master = disposition_all.merge(
    storage_agg, on=GRAIN, how="outer", suffixes=("", "_storage")
).merge(
    loss_agg, on=GRAIN, how="outer", suffixes=("", "_loss")
).merge(
    household_summary, on="household_id", how="left"
)

print(crop_loss_master.shape)
crop_loss_master.head()
```

```python
# Sanity checks
print("rows with a crop from disposition but no loss record:",
      crop_loss_master["total_loss_qty"].isna().sum())
print("rows with total_loss_qty > 0:", (crop_loss_master["total_loss_qty"] > 0).sum())
print("crop_domain counts:")
print(crop_loss_master["crop_domain"].value_counts(dropna=False))
crop_loss_master.describe(include="all").T
```

```python
OUT_PATH = Path("../data/processed/crop_loss_master_w1.csv")
crop_loss_master.to_csv(OUT_PATH, index=False)
print(f"saved {crop_loss_master.shape} to {OUT_PATH}")
```

## Caveats

- Columns suffixed `_unconfirmed` are structurally grouped (e.g. all 3 loss-reason triplets, all 6 disposition shares) but their exact real-world meaning (which reason, which disposition channel) is **not** verified — that requires the official LSMS-ISA Basic Information Document / questionnaire, which isn't in this repo yet.
- `region_code`/`zone_code`/`woreda_code` follow the standard published ERSS admin hierarchy convention but aren't decoded to actual place names (no lookup table available yet).
- `sect_nr_ph_w1` and `sect1_ph_w1` were intentionally excluded from the merge — they carry no crop or loss information (respondent-roster only).
- `crop_loss_master` is Wave 1 (2011-12) only. Repeat this pipeline per wave once Waves 2-5 are added to `data/raw/`, then concatenate before merging with CHIRPS rainfall.
