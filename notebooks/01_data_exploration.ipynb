# HarvestGuard — Data Exploration

Wave 1 (2011-12) LSMS-ISA Ethiopia data. Column names are coded (e.g. `ph_s9q01`) — no codebook is available yet, so this notebook explores structure and value ranges blind, without assuming what each column means. Goal: find the post-harvest loss variable(s) by inspection.

```python
import pandas as pd
from pathlib import Path

pd.set_option("display.max_columns", 100)

DATA_DIR = Path("../data/raw/ETH_2011_ERSS_v02_M_CSV")
sorted(p.name for p in DATA_DIR.glob("*.csv"))
```

## Post-harvest (`_ph_`) sections

These are the most likely candidates for loss-related data. Load each and inspect shape, dtypes, and a sample.

```python
ph_files = sorted(DATA_DIR.glob("*_ph_w1.csv"))
[p.name for p in ph_files]
```

```python
ph_frames = {p.stem: pd.read_csv(p) for p in ph_files}

for name, df in ph_frames.items():
    print(f"{name:25s} shape={df.shape}")
```

### Missingness across all PH sections

Columns that are mostly non-null and numeric with plausible quantity ranges are the best candidates for a loss/quantity variable.

```python
for name, df in ph_frames.items():
    print(f"\n--- {name} ---")
    print(df.isna().mean().sort_values().round(3))
```
